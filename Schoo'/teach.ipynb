```python
import random

player = 0
botScore = 0


def check(playerStore,bot,player=player,botScore=botScore):
    if playerStore in ["rock", "paper", "scissors"]:
        if playerStore == bot:
            print("DRAW")
        else:
            if playerStore == "rock" and bot == "paper":
                print("BOT WINS!")
                botScore+=1
            elif playerStore == "rock" and bot == "scissors":
                print("PLAYER WINS!")
                player+=1
            elif playerStore == "paper" and bot == "scissors":
                print("BOT WINS!")
                botScore+=1
            elif playerStore == "paper" and bot == "rock":
                print("PLAYER WINS!")
                player+=1
            elif playerStore == "scissors" and bot == "rock":
                print("BOT WINS!")
                botScore+=1
            elif playerStore == "scissors" and bot == "paper":
                print("PLAYER WINS!")
                player+=1
            
    return [player,botScore]
            
playerStore = input("rock paper of scissors?")
bot = random.choice(["rock","paper","scissors"])
print(bot)
x = check(playerStore=playerStore,bot=bot)

print(x[1])
```
